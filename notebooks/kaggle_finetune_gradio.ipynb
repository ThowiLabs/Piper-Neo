# Piper Neo Fine-tune — Kaggle reproducible

Este cuaderno instala una revisión fija del repositorio, crea un entorno Python 3.10 aislado y lanza la interfaz Gradio para preparar datasets, hacer fine-tune/resume y respaldar checkpoints en Hugging Face.

El entrenamiento usa `es-419` y 22,050 Hz por defecto. Fine-tune y Resume están separados para evitar heredar por error el epoch/global_step del modelo base.

```python
import os, subprocess
from pathlib import Path

REPO_URL = "https://github.com/ThowiLabs/Piper-Neo.git"
PINNED_REF = os.environ.get("PIPER_NEO_REF", "dac2511eb86ad61c18e32f627baa40e72da776a9")
ROOT = Path("/kaggle/working/Piper-Neo")

if not (ROOT / ".git").exists():
    subprocess.run(["git", "clone", REPO_URL, str(ROOT)], check=True)
else:
    subprocess.run(["git", "fetch", "--all", "--tags"], cwd=ROOT, check=True)

subprocess.run(["git", "checkout", "--force", PINNED_REF], cwd=ROOT, check=True)
subprocess.run(["bash", str(ROOT / "script/setup_kaggle_finetune.sh")], cwd=ROOT, check=True)
print("Revision activa:", subprocess.check_output(["git", "rev-parse", "HEAD"], cwd=ROOT, text=True).strip())
```

```python
import subprocess
PY = ROOT / "src/python/.venv/bin/python"
subprocess.run([str(PY), "-c", "import torch; print('CUDA:', torch.cuda.is_available()); print('GPU count:', torch.cuda.device_count())"], check=True)
```
